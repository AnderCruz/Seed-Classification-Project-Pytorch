# Explanation of seeds

This notebook explains each line of the code in the original `seeds`.

```python
# Cell 1: Import pandas library
import pandas as pd
```

## Explanation for Cell 1

This line imports the `pandas` library, which is widely used for data manipulation and analysis in Python. It provides data structures like DataFrames to handle tabular data efficiently.

```python
# Cell 2: Define URL for the dataset
url = 'https://raw.githubusercontent.com/alura-cursos/Primeiros_passos_pytorch/main/sementes.csv'
```

## Explanation for Cell 2

This line assigns a URL to the variable `url`. The URL points to a CSV file named `sementes.csv`, which contains the seed dataset used in this notebook.

```python
# Cell 3: Load dataset and display first rows
dados = pd.read_csv(url)
dados.head()
```

## Explanation for Cell 3

This cell reads the CSV file from the specified URL into a pandas DataFrame called `dados` using `pd.read_csv()`. The `dados.head()` method displays the first five rows of the DataFrame to provide a quick overview of the data, which includes features like area, perimeter, and species.

```python
# Cell 4: Create feature matrix
X = dados.drop(['Espécie'], axis=1).values
```

## Explanation for Cell 4

This line creates the feature matrix `X` by removing the 'Espécie' (species) column from the DataFrame using `drop()`. The `axis=1` parameter specifies that a column is being dropped. The `.values` attribute converts the resulting DataFrame into a NumPy array for further processing.

```python
# Cell 5: Create target vector
y = dados['Espécie'].values
```

## Explanation for Cell 5

This line extracts the 'Espécie' column from the DataFrame to create the target vector `y`, which contains the labels (species) for classification. The `.values` attribute converts it into a NumPy array.

```python
# Cell 6: Import train_test_split function
from sklearn.model_selection import train_test_split
```

## Explanation for Cell 6

This line imports the `train_test_split` function from the `sklearn.model_selection` module. This function is used to split the dataset into training and testing sets.

```python
# Cell 7: Split data into training and testing sets
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.2)
```

## Explanation for Cell 7

This line splits the feature matrix `X` and target vector `y` into training (`X_treino`, `y_treino`) and testing (`X_teste`, `y_teste`) sets. The `test_size=0.2` parameter indicates that 20% of the data is used for testing, and the remaining 80% for training.

```python
# Cell 8: Import PyTorch library
import torch
```

## Explanation for Cell 8

This line imports the `torch` library, which is the core of PyTorch, a popular framework for building and training neural networks.

```python
# Cell 9: Convert data to PyTorch tensors
X_treino = torch.FloatTensor(X_treino)
X_teste = torch.FloatTensor(X_teste)
y_treino = torch.LongTensor(y_treino)
y_teste = torch.LongTensor(y_teste)
```

## Explanation for Cell 9

This cell converts the training and testing data into PyTorch tensors. The features (`X_treino`, `X_teste`) are converted to `FloatTensor` for floating-point data, and the labels (`y_treino`, `y_teste`) are converted to `LongTensor` for integer-based class labels, as required by PyTorch for neural network processing.

```python
# Cell 10: Import PyTorch neural network modules
import torch.nn as nn
import torch.nn.functional as F
```

## Explanation for Cell 10

This cell imports the `nn` module for defining neural network layers and the `functional` module (`F`) for activation functions and other utilities from PyTorch.

```python
# Cell 11: Define neural network model
class Modelo(nn.Module):
    def __init__(self, entrada=7, camada_escondida1=14, camada_escondida2=49, saida=3):
        super().__init__()
        self.fc1 = nn.Linear(entrada, camada_escondida1)
        self.fc2 = nn.Linear(camada_escondida1, camada_escondida2)
        self.out = nn.Linear(camada_escondida2, saida)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.out(x)
        return x
```

## Explanation for Cell 11

This cell defines a neural network class `Modelo` that inherits from `nn.Module`. The `__init__` method initializes three fully connected layers:  
- `fc1`: A linear layer mapping 7 input features to 14 units in the first hidden layer.  
- `fc2`: A linear layer mapping 14 units to 49 units in the second hidden layer.  
- `out`: A linear layer mapping 49 units to 3 output units (for 3 classes).  
The `forward` method defines the forward pass, applying ReLU activation after the first two layers and returning the final output.

```python
# Cell 12: Instantiate the model
modelo_classificacao = Modelo()
modelo_classificacao
```

## Explanation for Cell 12

This cell creates an instance of the `Modelo` class called `modelo_classificacao` and prints its structure, showing the layers defined in the model.

```python
# Cell 13: Define loss function and optimizer
funcao_objetivo = nn.CrossEntropyLoss()
otimizador = torch.optim.Adam(modelo_classificacao.parameters(), lr=0.01)
```

## Explanation for Cell 13

This cell defines:  
- `funcao_objetivo`: The loss function as `CrossEntropyLoss`, suitable for multi-class classification.  
- `otimizador`: The Adam optimizer with a learning rate of 0.01, which will update the model's parameters during training.

```python
# Cell 14: Set training parameters
epocas = 100
custos = []
```

## Explanation for Cell 14

This cell sets the number of training epochs to 100 and initializes an empty list `custos` to store the loss values for each epoch.

```python
# Cell 15: Training loop
for i in range(epocas):
    y_predito = modelo_classificacao.forward(X_treino)
    custo = funcao_objetivo(y_predito, y_treino)
    custos.append(custo)
    
    otimizador.zero_grad()
    custo.backward()
    otimizador.step()
```

## Explanation for Cell 15

This cell implements the training loop for 100 epochs:  
- `y_predito = modelo_classificacao.forward(X_treino)`: Performs a forward pass on the training data to get predictions.  
- `custo = funcao_objetivo(y_predito, y_treino)`: Computes the loss between predictions and true labels.  
- `custos.append(custo)`: Stores the loss value.  
- `otimizador.zero_grad()`: Resets the gradients to zero to avoid accumulation.  
- `custo.backward()`: Computes gradients via backpropagation.  
- `otimizador.step()`: Updates the model parameters using the optimizer.

```python
# Cell 16: Make predictions on test set
preds = []
with torch.no_grad():
    for val in X_teste:
        y_predito = modelo_classificacao.forward(val)
        preds.append(y_predito.argmax().item())
```

## Explanation for Cell 16

This cell generates predictions for the test set:  
- `preds = []`: Initializes an empty list to store predictions.  
- `with torch.no_grad()`: Disables gradient computation for efficiency during inference.  
- The loop iterates over each test sample, performs a forward pass, and appends the predicted class (using `argmax().item()`) to `preds`.

```python
# Cell 17: Create DataFrame with predictions and accuracy
df = pd.DataFrame({'Y': y_teste, 'YHat': preds})
df['Correto'] = [1 if corr == pred else 0 for corr, pred in zip(df['Y'], df['YHat'])]
df
```

## Explanation for Cell 17

This cell creates a pandas DataFrame to compare true and predicted labels:  
- `df = pd.DataFrame({'Y': y_teste, 'YHat': preds})`: Creates a DataFrame with true labels (`Y`) and predicted labels (`YHat`).  
- `df['Correto'] = [1 if corr == pred else 0 ...]`: Adds a 'Correto' column, where 1 indicates a correct prediction and 0 indicates an incorrect one, using a list comprehension with `zip`.  
- `df`: Displays the resulting DataFrame.
